```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_values = [20, 50, 100, 200, 500]
M = 2000
seed = 450
samples = {}
empirical_means = {}
binom_param = 0.3

np.random.seed(seed)

for n in n_values:
    samples[n] = []
    for i in range(M):
        sample_normalized_mean = (np.random.binomial(n, binom_param) / n) - binom_param
        samples[n].append(sample_normalized_mean)
    empirical_means[n] = np.mean(np.abs(samples[n]))
```

```python
n_smooth = np.linspace(min(n_values), max(n_values), 1000)
rate = 1 / np.sqrt(n_smooth)

plt.figure(figsize=(8, 5))

plt.plot(
    n_values,
    list(empirical_means.values()),
    "o-",
    label="Empirical mean absolute deviation"
)

plt.plot(
    n_smooth,
    rate,
    "--",
    label=r"$n^{-1/2}$"
)

plt.xlabel("Sample size (n)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("figures/question 1/1c_convergence_rate.png", dpi=300, bbox_inches="tight")
plt.xscale("log")
plt.yscale("log")
plt.savefig("figures/question 1/1c_convergence_rate_loglog.png", dpi=300, bbox_inches="tight")
plt.show()
```
